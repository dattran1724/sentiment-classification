```python
import re
import pandas as pd
import os


input_path = r"D:\2025-2026\ChuyenDe\New_data\csv\test_gold\test_gold.csv"

output_folder = os.path.dirname(input_path) 
output_path = os.path.join(output_folder, "test_gold_final.csv")

try:
    df = pd.read_csv(input_path)
    print(f"Kích thước ban đầu: {df.shape} (Hàng, Cột)")
except FileNotFoundError:
    print("Lỗi: Không tìm thấy file. Vui lòng kiểm tra lại đường dẫn.")
    exit()

null_count = df.isnull().sum().sum() 

if null_count > 0:
    rows_before = df.shape[0]
    
    # Xóa hàng có ít nhất 1 giá trị null
    df.dropna(inplace=True)
    df.reset_index(drop=True, inplace=True)
    
    rows_after = df.shape[0]
    print(f"Phát hiện dữ liệu lỗi. Đã xóa {rows_before - rows_after} hàng chứa Null.")
else:
    print("Dữ liệu không có giá trị Null nào.")

def clean_text(text: str) -> str:
    if isinstance(text, list) and len(text) > 0:
        text = text[0]
    elif not isinstance(text, str):
        text = str(text)

    text = re.sub(r'<.*?>', '', text)

    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    emoji_pattern = re.compile("["  
        u"\U0001F600-\U0001F64F"  
        u"\U0001F300-\U0001F5FF"    
        u"\U0001F680-\U0001F6FF"   
        u"\U0001F1E0-\U0001F1FF"  
        u"\U00002702-\U000027B0"  
        u"\U000024C2-\U0001F251"  
        "]+", flags=re.UNICODE)
    text = emoji_pattern.sub('', text)

    text = re.sub(r'[^\w\s]', '', text, flags=re.UNICODE)

    text = re.sub(r'\s+', ' ', text.strip())

    return text

df.to_csv(output_path, index=False, encoding='utf-8-sig')

print(f"Kích thước cuối cùng: {df.shape}")
print(f"Đã lưu file thành công tại: {output_path}")
```

```python
import pandas as pd

# =========================
# 1. LUẬT GỘP NHÃN
# =========================
MERGE_RULES = {
    # SOFTWARE
    "SOFTWARE#PRICE": "SOFTWARE#USABILITY",
    "SOFTWARE#QUALITY": "SOFTWARE#USABILITY",

    # SERVICE
    "SERVICE#PERFORMANCE": "SERVICE#QUALITY",
    "SERVICE#PRICE": "SERVICE#QUALITY",
    "SERVICE#USABILITY": "SERVICE#QUALITY",

    # INPUT_DEVICES
    "INPUT_DEVICES#DESIGN": "INPUT_DEVICES#GENERAL",
    "INPUT_DEVICES#PERFORMANCE": "INPUT_DEVICES#GENERAL",
    "INPUT_DEVICES#QUALITY": "INPUT_DEVICES#GENERAL",
    "INPUT_DEVICES#USABILITY": "INPUT_DEVICES#GENERAL",

    # DISPLAY
    "DISPLAY#PERFORMANCE": "DISPLAY#GENERAL",
    "DISPLAY#USABILITY": "DISPLAY#GENERAL",

    # CAMERA
    "CAMERA#USABILITY": "CAMERA#GENERAL",
    "CAMERA#PRICE": "CAMERA#GENERAL",
}

# =========================
# 2. NHÃN XOÁ → XOÁ LUÔN HÀNG
# =========================
DROP_LABELS = {
    # HARDWARE
    "HARDWARE#GENERAL",
    "HARDWARE#PRICE",
    "HARDWARE#USABILITY",

    # GRAPHICS
    "GRAPHICS#DESIGN",
    "GRAPHICS#GENERAL",
    "GRAPHICS#QUALITY",

    # FOOD_DRINKS
    "FOOD_DRINKS#GENERAL",

    # BATTERY
    "BATTERY#QUALITY",
    "BATTERY#USABILITY",
    "BATTERY#GENERAL",
    "BATTERY#DESIGN",
}

# =========================
# 3. RELABEL + DROP ROWS
# =========================
# df phải có cột 'category'

# Bước 1: gộp nhãn
df["category"] = df["category"].replace(MERGE_RULES)

# Bước 2: xoá LUÔN các hàng có nhãn không dùng
before = len(df)
df = df[~df["category"].isin(DROP_LABELS)].reset_index(drop=True)
after = len(df)

# =========================
# 4. KIỂM TRA
# =========================
print(f"Số hàng bị xoá: {before - after}")
print(f"Số hàng còn lại: {after}")
print(f"Số nhãn còn lại: {df['category'].nunique()}")

print("\nPhân bố category sau khi xử lý:")
counts = df["category"].value_counts()
print(counts[counts > 100].sort_values(ascending=False))

output_dir = os.path.dirname(input_path)
output_path = os.path.join(output_dir, "train_final.csv")
df.to_csv(output_path, index=False)

```
