```python
import pandas as pd

path = r'D:\2025-2026\ChuyenDe\New_data\csv\train\train_final.csv'
df_aug = pd.read_csv(path)

category_counts = df_aug['category'].value_counts().sort_index()
    
print(f"Tổng số nhãn duy nhất: {len(category_counts)}")
print("Class distribution:")
print(df_aug['category'].value_counts())

polarity_counts = df_aug['polarity'].value_counts().sort_index()
    
print(f"\n\nTổng số nhãn duy nhất: {len(polarity_counts)}")
print("Class distribution:")
print(df_aug['polarity'].value_counts())

language_counts = df_aug['language'].value_counts().sort_index()
    
print(f"\n\nTổng số nhãn duy nhất: {len(language_counts)}")
print("Class distribution:")
print(df_aug['language'].value_counts())
```
